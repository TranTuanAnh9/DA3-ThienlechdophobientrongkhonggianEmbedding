# 08 — Liên hệ Thiên lệch Đầu ra với Hình học Embedding & Kiểm tra Ổn định qua 2 Seed

Notebook này kết nối hai góc nhìn của **SV1 (Popularity Bias ở đầu ra)** và **SV2 (Hình học không gian Embedding)** trên cùng bảng phân tích cấp độ item (`results/item_level/*.parquet`), đồng thời kiểm tra tính ổn định của mọi kết luận qua **2 random seed (`2024` và `2025`)**.

## Nội dung phân tích chính (Mục 12 & Mục 15 tài liệu đề tài)

1. **Ba tương quan cốt lõi (Spearman $\rho$, p-value, 95% Bootstrap CI):**
   - `train_freq ↔ embedding_norm`: Item phổ biến trong train có vector dài hơn không?
   - `train_freq ↔ rec_count`: Item phổ biến trong train có được gợi ý nhiều hơn trong Top-20 không?
   - `embedding_norm ↔ rec_count`: Item có norm lớn hơn có được gợi ý nhiều hơn không?
2. **Kiểm soát biến gây nhiễu `train_frequency`:**
   - Tính tương quan `embedding_norm ↔ rec_count` bên trong từng bin tần suất (`rare`, `medium`, `popular`).
   - Hồi quy đa biến $\log(1 + \text{rec\_count}_i) = \beta_0 + \beta_1 \log(1 + \text{train\_freq}_i) + \beta_2 \lVert e_i \rVert_2$.
3. **Kiểm soát biến gây nhiễu bằng standardized beta, khoảng tin cậy 95% và tương quan trong từng bin.**
4. **Kết nối accuracy, popularity bias, cosine/category centroid và cold item trong cùng bảng item-level.**
5. **Kiểm tra độ ổn định qua 2 seed (`seed_stability.csv`) và bảng tổng hợp cuối (`final_model_comparison.csv`).**

> Lưu ý: đây là phân tích quan sát. Tương quan và hồi quy không chứng minh embedding norm là nguyên nhân gây popularity bias.

### Đầu ra chính
- Bảng: `bias_geometry_relationship.csv`, `seed_stability.csv`, `final_model_comparison.csv`, `final_key_results.csv`, `final_conclusions.csv`.
- Kiểm định: `within_bin_correlations.csv`, `ols_regression_results.csv`, các bảng ổn định qua seed.
- Hình: frequency–recommendation, norm–recommendation, within-bin, standardized beta và bảng so sánh cuối.

## 1. Nạp dữ liệu từ `results/item_level/` và `results/tables/`

```python
from pathlib import Path
import sys
import warnings

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

warnings.filterwarnings("ignore", category=FutureWarning)
pd.set_option("display.max_columns", 50)
pd.set_option("display.float_format", lambda v: f"{v:,.4f}")
sns.set_theme(style="whitegrid", context="notebook")


def find_project_root(start: Path | None = None) -> Path:
    current = (start or Path.cwd()).resolve()
    for candidate in (current, *current.parents):
        if (candidate / "BTL-K03-tong-quan-de-tai.md").exists():
            return candidate
    raise FileNotFoundError("Không tìm thấy thư mục gốc dự án BTL-K03.")


PROJECT_ROOT = find_project_root()
if str(PROJECT_ROOT / "src") not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT / "src"))

from btl_k03.paths import (
    FIGURES_DIR, ITEM_LEVEL_DIR, PROCESSED_DIGI_DIR, RUNS_DIR,
    STAT_TESTS_DIR, TABLES_DIR, ensure_output_dirs,
)
from btl_k03.embedding_analysis import (
    analyze_three_core_relationships,
    attach_item_accuracy,
    build_query_and_item_accuracy_tables,
    compute_geometry_output_correlations,
    fit_geometry_accuracy_models,
    fit_geometry_exposure_models,
    summarize_geometry_and_outcomes_by_group,
)
from btl_k03.utils import load_experiment_config

ensure_output_dirs()

COMMON_CFG = load_experiment_config("gru4rec")
MODELS = ["gru4rec", "narm", "sasrec"]
SEEDS = [int(seed) for seed in COMMON_CFG["seeds"]]
EXPECTED_PAIRS = {(m.upper(), s) for m in MODELS for s in SEEDS}

required_table_paths = {
    "accuracy": TABLES_DIR / "model_accuracy.csv",
    "bias": TABLES_DIR / "popularity_bias.csv",
    "cosine": TABLES_DIR / "cosine_category.csv",
    "centroid": TABLES_DIR / "category_centroid.csv",
}
missing_tables = [str(path) for path in required_table_paths.values() if not path.is_file()]
if missing_tables:
    raise FileNotFoundError("Thiếu đầu vào từ notebook 06/07: " + ", ".join(missing_tables))

required_item_columns = {
    "model", "seed", "internal_item_id", "item_id", "train_freq",
    "popularity_group", "is_trainable_warm", "embedding_norm",
    "rec_count", "exposure_share", "distance_to_category_centroid",
    "cosine_to_category_centroid",
}
item_frames = []
integrity_rows = []
for m in MODELS:
    for s in SEEDS:
        path = ITEM_LEVEL_DIR / f"{m}_seed{s}.parquet"
        if not path.is_file():
            raise FileNotFoundError(f"Thiếu {path}. Hãy chạy các notebook 03, 04, 05 trước.")
        frame = pd.read_parquet(path).sort_values("internal_item_id").reset_index(drop=True)
        missing_columns = sorted(required_item_columns.difference(frame.columns))
        if missing_columns:
            raise ValueError(f"{path.name} thiếu cột: {missing_columns}")
        if frame[["train_freq", "embedding_norm", "rec_count", "exposure_share"]].isna().any().any():
            raise ValueError(f"{path.name} chứa giá trị thiếu ở các biến phân tích chính.")
        if not np.isfinite(frame[["train_freq", "embedding_norm", "rec_count", "exposure_share"]]).all().all():
            raise ValueError(f"{path.name} chứa NaN/Inf.")
        if set(frame["model"].unique()) != {m} or set(frame["seed"].unique()) != {s}:
            raise ValueError(f"Nhãn model/seed không khớp trong {path.name}.")
        item_frames.append(frame)
        integrity_rows.append({
            "model": m.upper(), "seed": s, "n_items": len(frame),
            "n_warm": int(frame["is_trainable_warm"].sum()),
            "recommendation_slots": int(frame["rec_count"].sum()),
            "zero_rec_warm_pct": 100 * float((frame.loc[frame["is_trainable_warm"], "rec_count"] == 0).mean()),
            "status": "OK",
        })

all_item_df = pd.concat(item_frames, ignore_index=True)
integrity_df = pd.DataFrame(integrity_rows)
accuracy_df = pd.read_csv(required_table_paths["accuracy"])
bias_df = pd.read_csv(required_table_paths["bias"])
cosine_df = pd.read_csv(required_table_paths["cosine"])
centroid_df = pd.read_csv(required_table_paths["centroid"])
centroid_df["model"] = centroid_df["model"].str.upper()

def assert_model_seed_pairs(df: pd.DataFrame, name: str) -> None:
    pairs = set(map(tuple, df[["model", "seed"]].drop_duplicates().to_numpy()))
    if pairs != EXPECTED_PAIRS:
        raise ValueError(f"{name} có model/seed={pairs}, kỳ vọng={EXPECTED_PAIRS}.")

for name, frame in [("accuracy", accuracy_df), ("bias", bias_df), ("cosine", cosine_df), ("centroid", centroid_df)]:
    assert_model_seed_pairs(frame, name)
if accuracy_df.duplicated(["model", "seed"]).any() or bias_df.duplicated(["model", "seed"]).any():
    raise ValueError("Bảng accuracy/bias có dòng model-seed bị trùng.")
if not (integrity_df["n_items"].nunique() == integrity_df["n_warm"].nunique() == integrity_df["recommendation_slots"].nunique() == 1):
    raise ValueError("Catalog hoặc số recommendation slots không nhất quán giữa các lần chạy.")

print(f"Đã nạp và kiểm tra {len(all_item_df):,} dòng item-level từ 3 mô hình × 2 seed.")
display(integrity_df)
```

## 2. Phân tích 3 tương quan cốt lõi, tương quan trong từng Frequency Bin và Hồi quy đa biến

```python
N_BOOTSTRAP = int(COMMON_CFG["embedding_geometry"].get("final_analysis_bootstrap_iterations", 500))
core_corr_frames = []
bin_corr_frames = []
reg_rows = []

for (m, s), group_df in all_item_df.groupby(["model", "seed"], sort=False):
    c_df, b_df, r_dict = analyze_three_core_relationships(
        group_df, n_bootstrap=N_BOOTSTRAP, seed=int(s)
    )
    core_corr_frames.append(c_df)
    bin_corr_frames.append(b_df)
    reg_rows.append(r_dict)

core_corr_df = pd.concat(core_corr_frames, ignore_index=True)
bin_corr_df = pd.concat(bin_corr_frames, ignore_index=True)
reg_df = pd.DataFrame(reg_rows)

# Tổng hợp mean ± std qua hai seed cho ba lớp bằng chứng.
warm_core = core_corr_df.loc[core_corr_df["scope"] == "warm_only"].copy()
core_seed_summary = (
    warm_core.groupby(["model", "pair"], sort=False)
    .agg(n_seeds=("seed", "nunique"), rho_mean=("rho", "mean"), rho_std=("rho", "std"),
         rho_min=("rho", "min"), rho_max=("rho", "max"))
    .reset_index()
)
bin_seed_summary = (
    bin_corr_df.groupby(["model", "popularity_group"], sort=False)
    .agg(n_seeds=("seed", "nunique"), norm_rec_rho_mean=("norm_rec_rho", "mean"),
         norm_rec_rho_std=("norm_rec_rho", "std"),
         freq_norm_rho_mean=("freq_norm_rho", "mean"))
    .reset_index()
)
reg_seed_summary = (
    reg_df.groupby("model", sort=False)
    .agg(
        n_seeds=("seed", "nunique"),
        freq_beta_std_mean=("log1p_freq_beta1_standardized", "mean"),
        freq_beta_std_std=("log1p_freq_beta1_standardized", "std"),
        norm_beta_std_mean=("norm_beta2_standardized", "mean"),
        norm_beta_std_std=("norm_beta2_standardized", "std"),
        norm_beta_raw_mean=("norm_beta2", "mean"),
        r_squared_mean=("r_squared", "mean"),
        r_squared_std=("r_squared", "std"),
    )
    .reset_index()
)

core_corr_df.to_csv(TABLES_DIR / "bias_geometry_relationship.csv", index=False)
bin_corr_df.to_csv(STAT_TESTS_DIR / "within_bin_correlations.csv", index=False)
reg_df.to_csv(STAT_TESTS_DIR / "ols_regression_results.csv", index=False)
core_seed_summary.to_csv(TABLES_DIR / "core_relationship_seed_summary.csv", index=False)
bin_seed_summary.to_csv(STAT_TESTS_DIR / "within_bin_seed_summary.csv", index=False)
reg_seed_summary.to_csv(STAT_TESTS_DIR / "regression_seed_summary.csv", index=False)

print("=== BẢNG 1: BA TƯƠNG QUAN SPEARMAN CỐT LÕI (FREQ ↔ NORM, FREQ ↔ REC, NORM ↔ REC) ===")
display(core_corr_df.loc[core_corr_df["scope"] == "warm_only"])

print("\n=== BẢNG 2: TƯƠNG QUAN NORM ↔ REC_COUNT TRONG TỪNG NHÓM TẦN SUẤT (KIỂM SOÁT FREQUENCY) ===")
display(bin_corr_df)

print("\n=== BẢNG 3: HỒI QUY ĐA BIẾN log(1 + rec_count) = β0 + β1*log(1 + train_freq) + β2*norm ===")
display(reg_df)

print("\n=== BẢNG 4: STANDARDIZED BETA VÀ ĐỘ ỔN ĐỊNH QUA HAI SEED ===")
display(reg_seed_summary)
```

## 3. Bảng Tổng hợp So sánh Cuối cùng & Kiểm tra Độ ổn định qua 2 Seed

```python
# 1. Ba tương quan chính về dạng cột cho từng (model, seed).
warm_corrs = core_corr_df.loc[core_corr_df["scope"] == "warm_only"].copy()
warm_corrs["model"] = warm_corrs["model"].str.upper()
piv_corr = warm_corrs.pivot(index=["model", "seed"], columns="pair", values="rho").reset_index()
piv_corr = piv_corr.rename(columns={
    "train_freq <-> embedding_norm": "spearman_freq_norm",
    "train_freq <-> rec_count": "spearman_freq_rec",
    "embedding_norm <-> rec_count": "spearman_norm_rec",
})

# 2. Cosine cùng/khác category và hai khoảng chênh có ý nghĩa diễn giải.
piv_cos = cosine_df.pivot(index=["model", "seed"], columns="pair_group", values="mean_cosine").reset_index()
required_cosine_groups = {"warm_same_category", "warm_diff_category", "cold_warm_same_category", "cold_warm_diff_category"}
if not required_cosine_groups.issubset(piv_cos.columns):
    raise ValueError(f"Thiếu nhóm cosine: {sorted(required_cosine_groups.difference(piv_cos.columns))}")
piv_cos["warm_category_cosine_gap"] = piv_cos["warm_same_category"] - piv_cos["warm_diff_category"]
piv_cos["cold_category_cosine_gap"] = piv_cos["cold_warm_same_category"] - piv_cos["cold_warm_diff_category"]

# 3. Khoảng cách tới category centroid theo popular/medium/rare/cold.
piv_centroid = centroid_df.pivot(
    index=["model", "seed"], columns="popularity_group", values="distance_mean"
).reset_index().rename(columns={
    "popular": "centroid_distance_popular", "medium": "centroid_distance_medium",
    "rare": "centroid_distance_rare", "cold": "centroid_distance_cold",
})
piv_centroid["centroid_rare_minus_popular"] = (
    piv_centroid["centroid_distance_rare"] - piv_centroid["centroid_distance_popular"]
)
piv_centroid["centroid_cold_minus_popular"] = (
    piv_centroid["centroid_distance_cold"] - piv_centroid["centroid_distance_popular"]
)

# 4. Tương quan norm-rec_count trong từng popularity bin.
piv_bin = bin_corr_df.pivot(
    index=["model", "seed"], columns="popularity_group", values="norm_rec_rho"
).reset_index().rename(columns={
    "rare": "norm_rec_rho_rare", "medium": "norm_rec_rho_medium",
    "popular": "norm_rec_rho_popular",
})
piv_bin["model"] = piv_bin["model"].str.upper()

# 5. Hồi quy kiểm soát frequency, gồm beta thô và standardized beta.
reg_for_merge = reg_df.copy()
reg_for_merge["model"] = reg_for_merge["model"].str.upper()
reg_columns = [
    "model", "seed", "log1p_freq_beta1_standardized",
    "norm_beta2", "norm_p", "norm_ci95_low", "norm_ci95_high",
    "norm_beta2_standardized", "norm_standardized_ci95_low",
    "norm_standardized_ci95_high", "r_squared", "adjusted_r_squared",
    "zero_rec_count_share",
]

full_run_table = (
    accuracy_df[["model", "seed", "recall@20", "mrr@20", "ndcg@20", "last_item_recall@20", "full_test_effective_recall@20"]]
    .merge(
        bias_df[["model", "seed", "coverage_warm@20", "gini_warm@20", "arp@20",
                 "popular_group_exposure_share_pct", "rare_group_exposure_share_pct"]],
        on=["model", "seed"], validate="one_to_one",
    )
    .merge(piv_corr, on=["model", "seed"], validate="one_to_one")
    .merge(piv_cos, on=["model", "seed"], validate="one_to_one")
    .merge(piv_centroid, on=["model", "seed"], validate="one_to_one")
    .merge(piv_bin, on=["model", "seed"], validate="one_to_one")
    .merge(reg_for_merge[reg_columns], on=["model", "seed"], validate="one_to_one")
)

actual_pairs = set(map(tuple, full_run_table[["model", "seed"]].to_numpy()))
if actual_pairs != EXPECTED_PAIRS or len(full_run_table) != len(EXPECTED_PAIRS):
    raise ValueError(f"Bảng tổng hợp bị thiếu model/seed: {actual_pairs}")
if full_run_table.isna().any().any():
    missing = full_run_table.columns[full_run_table.isna().any()].tolist()
    raise ValueError(f"Bảng tổng hợp có giá trị thiếu ở: {missing}")

full_run_table.to_csv(TABLES_DIR / "seed_stability.csv", index=False)
print("=== BẢNG 5: CHI TIẾT TỪNG MODEL × SEED ===")
display(full_run_table)

final_comparison = full_run_table.drop(columns=["seed"]).groupby("model", sort=False).agg(["mean", "std"])
final_comparison.columns = [f"{col}_{stat}" for col, stat in final_comparison.columns]
final_comparison = final_comparison.reset_index()
final_comparison.to_csv(TABLES_DIR / "final_model_comparison.csv", index=False)

key_metrics = [
    "recall@20", "coverage_warm@20", "gini_warm@20", "arp@20",
    "popular_group_exposure_share_pct", "rare_group_exposure_share_pct",
    "spearman_freq_norm", "spearman_freq_rec", "spearman_norm_rec",
    "norm_rec_rho_rare", "norm_rec_rho_medium", "norm_rec_rho_popular",
    "norm_beta2_standardized", "r_squared", "warm_category_cosine_gap",
    "cold_category_cosine_gap", "centroid_rare_minus_popular",
    "centroid_cold_minus_popular",
]
final_key_results = full_run_table.groupby("model", sort=False)[key_metrics].agg(["mean", "std"])
final_key_results.columns = [f"{col}_{stat}" for col, stat in final_key_results.columns]
final_key_results = final_key_results.reset_index()
final_key_results.to_csv(TABLES_DIR / "final_key_results.csv", index=False)

model_diagnostic = full_run_table.groupby("model", sort=False)[key_metrics].mean().reset_index()
model_diagnostic["accuracy_rank"] = model_diagnostic["recall@20"].rank(ascending=False, method="min").astype(int)
model_diagnostic["coverage_rank"] = model_diagnostic["coverage_warm@20"].rank(ascending=False, method="min").astype(int)
model_diagnostic["low_gini_rank"] = model_diagnostic["gini_warm@20"].rank(ascending=True, method="min").astype(int)
model_diagnostic.to_csv(TABLES_DIR / "model_diagnostic_summary.csv", index=False)

print("\n=== BẢNG 6: TỔNG HỢP CUỐI CÙNG MEAN ± STD QUA HAI SEED ===")
display(final_key_results)
print("\n=== BẢNG 7: CHẨN ĐOÁN/RANKING BA MÔ HÌNH ===")
display(model_diagnostic)
```

## 4. Biểu đồ Liên hệ giữa Hình học Embedding và Tần suất được Gợi ý

```python
first_seed = SEEDS[0]
group_colors = {"popular": "#d95f02", "medium": "#7570b3", "rare": "#1b9e77"}

def plot_item_relationship(seed: int, relation: str) -> None:
    fig, axes = plt.subplots(1, 3, figsize=(18, 5.2))
    for idx, m in enumerate(MODELS):
        sub = all_item_df.loc[
            (all_item_df["model"] == m) & (all_item_df["seed"] == seed) & all_item_df["is_trainable_warm"]
        ].copy()
        sub["log10_train_freq"] = np.log10(sub["train_freq"])
        sub["log10_rec_count"] = np.log10(1.0 + sub["rec_count"])
        sampled = sub.sample(n=min(6000, len(sub)), random_state=seed)
        row = full_run_table.loc[(full_run_table["model"] == m.upper()) & (full_run_table["seed"] == seed)].iloc[0]
        if relation == "frequency_recommendation":
            x_col, x_label = "log10_train_freq", "log10(Train Frequency)"
            rho, heading = row["spearman_freq_rec"], "Frequency vs Recommendation Count"
        else:
            x_col, x_label = "embedding_norm", "L2 Norm ||e_i||_2"
            rho, heading = row["spearman_norm_rec"], "Norm vs Recommendation Count"
        sns.scatterplot(
            data=sampled, x=x_col, y="log10_rec_count", hue="popularity_group",
            hue_order=["popular", "medium", "rare"], palette=group_colors,
            alpha=0.42, s=18, ax=axes[idx],
        )
        axes[idx].set_title(f"{m.upper()} — {heading}\nSpearman ρ={rho:.4f}, seed={seed}")
        axes[idx].set_xlabel(x_label)
        axes[idx].set_ylabel("log10(1 + Recommendation Count)")
    plt.tight_layout()
    seed_path = FIGURES_DIR / f"{relation}_seed{seed}.png"
    plt.savefig(seed_path, dpi=300)
    if seed == first_seed:
        plt.savefig(FIGURES_DIR / f"{relation}.png", dpi=300)
    plt.show()

for seed in SEEDS:
    plot_item_relationship(seed, "frequency_recommendation_relationship")
    plot_item_relationship(seed, "norm_recommendation_relationship")

# Phân bố recommendation count theo popularity group (thang log để tránh long-tail che khuất).
box_df = all_item_df.loc[(all_item_df["seed"] == first_seed) & all_item_df["is_trainable_warm"]].copy()
box_df["model_upper"] = box_df["model"].str.upper()
box_df["log10_rec_count"] = np.log10(1.0 + box_df["rec_count"])
fig, ax = plt.subplots(figsize=(11, 5.8))
sns.boxplot(
    data=box_df, x="popularity_group", y="log10_rec_count", hue="model_upper",
    order=["rare", "medium", "popular"], palette="Set2", showfliers=False, ax=ax,
)
ax.set_title(f"Recommendation Count theo Nhóm Popularity — Seed {first_seed}")
ax.set_xlabel("Nhóm item")
ax.set_ylabel("log10(1 + Recommendation Count)")
ax.legend(title="Mô hình")
plt.tight_layout()
plt.savefig(FIGURES_DIR / "recommendation_count_by_group.png", dpi=300)
plt.show()

# Quan hệ norm-rec_count sau khi chia theo frequency bin.
bin_plot_df = bin_corr_df.copy()
bin_plot_df["model"] = bin_plot_df["model"].str.upper()
fig, ax = plt.subplots(figsize=(10.5, 5.6))
sns.barplot(
    data=bin_plot_df, x="popularity_group", y="norm_rec_rho", hue="model",
    order=["rare", "medium", "popular"], palette="Set2", errorbar="sd", ax=ax,
)
ax.axhline(0, color="black", linewidth=1)
ax.set_title("Spearman(Norm, Recommendation Count) trong từng Frequency Bin")
ax.set_xlabel("Nhóm popularity")
ax.set_ylabel("Spearman ρ — mean ± SD qua 2 seed")
plt.tight_layout()
plt.savefig(FIGURES_DIR / "within_bin_norm_recommendation.png", dpi=300)
plt.show()

# Standardized beta của norm sau khi kiểm soát log(1 + train frequency), kèm CI 95%.
reg_plot_df = reg_df.copy()
reg_plot_df["model"] = reg_plot_df["model"].str.upper()
model_x = {model: idx for idx, model in enumerate([m.upper() for m in MODELS])}
seed_offsets = {seed: offset for seed, offset in zip(SEEDS, [-0.10, 0.10])}
seed_colors = {SEEDS[0]: "#66c2a5", SEEDS[1]: "#fc8d62"}
fig, ax = plt.subplots(figsize=(9.5, 5.6))
for seed in SEEDS:
    sub = reg_plot_df.loc[reg_plot_df["seed"] == seed].copy()
    sub["x"] = sub["model"].map(model_x) + seed_offsets[seed]
    yerr = np.vstack([
        sub["norm_beta2_standardized"] - sub["norm_standardized_ci95_low"],
        sub["norm_standardized_ci95_high"] - sub["norm_beta2_standardized"],
    ])
    ax.errorbar(
        sub["x"], sub["norm_beta2_standardized"], yerr=yerr, fmt="o",
        capsize=5, markersize=7, color=seed_colors[seed], label=f"Seed {seed}",
    )
ax.axhline(0, color="black", linewidth=1)
ax.set_xticks(range(len(MODELS)), [m.upper() for m in MODELS])
ax.set_title("Ảnh hưởng riêng của Norm sau khi Kiểm soát Train Frequency")
ax.set_xlabel("Mô hình")
ax.set_ylabel("Standardized beta2 của Embedding Norm (95% CI)")
ax.legend()
plt.tight_layout()
plt.savefig(FIGURES_DIR / "regression_standardized_coefficients.png", dpi=300)
plt.show()

# Sáu chỉ số nối accuracy, output bias và geometry trên cùng một hình tổng hợp.
fig, axes = plt.subplots(2, 3, figsize=(17, 10))
metrics_to_plot = [
    ("recall@20", "Recall@20 ↑"),
    ("coverage_warm@20", "Catalog Coverage@20 ↑"),
    ("gini_warm@20", "Recommendation Gini@20 ↓"),
    ("spearman_freq_norm", "Spearman: Frequency ↔ Norm"),
    ("spearman_freq_rec", "Spearman: Frequency ↔ Rec Count"),
    ("spearman_norm_rec", "Spearman: Norm ↔ Rec Count"),
]
for ax, (col, title) in zip(axes.flat, metrics_to_plot):
    sns.barplot(data=full_run_table, x="model", y=col, hue="seed", palette="Set2", errorbar=None, ax=ax)
    for container in ax.containers:
        ax.bar_label(container, fmt="%.3f", padding=2, fontsize=8)
    ax.set_title(title)
    ax.set_xlabel("Mô hình")
    ax.set_ylabel(col)
plt.tight_layout()
plt.savefig(FIGURES_DIR / "final_model_comparison.png", dpi=300)
plt.show()
```

## 5. Nối trực tiếp hình học embedding với exposure và accuracy

Phần này là cầu nối chính giữa notebook 06 và 07. Ta kiểm tra hai outcome khác nhau:

- **Exposure cấp item:** một item xuất hiện bao nhiêu lần trong toàn bộ Top-20 của 40.001 truy vấn test warm.
- **Accuracy cấp truy vấn:** target thật của từng truy vấn có xuất hiện trong Top-20 hay không (`hit@20`).

Ba đặc trưng hình học được nối với hai outcome là norm, khoảng cách tới tâm category và cosine với tâm category. Phân tích gồm ba tầng: (1) so sánh popular/medium/rare; (2) Spearman toàn warm catalog và trong từng nhóm; (3) OLS/logistic regression có kiểm soát `log(1 + train_frequency)`.

> Cold item không được đưa vào hồi quy outcome chính vì test chính là test warm và cold embedding không được học từ interaction train. Kết quả chỉ thể hiện **mối liên hệ**, không chứng minh quan hệ nhân quả.

```python
test_examples = pd.read_parquet(PROCESSED_DIGI_DIR / "test_examples.parquet")
if len(test_examples) != 40_001:
    raise ValueError("Test warm phải có đúng 40.001 truy vấn all-prefix.")

enriched_item_frames, query_accuracy_frames = [], []
geometry_group_frames, geometry_corr_frames = [], []
geometry_exposure_reg_frames, geometry_accuracy_reg_frames = [], []
for model in MODELS:
    for seed in SEEDS:
        item_frame = all_item_df.loc[(all_item_df["model"] == model) & (all_item_df["seed"] == seed)].copy()
        top20_path = RUNS_DIR / model / f"seed_{seed}" / "top20.parquet"
        if not top20_path.is_file():
            raise FileNotFoundError(f"Thiếu {top20_path}.")
        top20 = pd.read_parquet(top20_path)
        query_accuracy, item_accuracy = build_query_and_item_accuracy_tables(test_examples, top20, topk=20)
        if len(query_accuracy) != 40_001 or len(top20) != 800_020:
            raise ValueError(f"{model}-seed{seed}: sai số truy vấn hoặc recommendation slots.")
        enriched = attach_item_accuracy(item_frame, item_accuracy, topk=20)
        query_accuracy["model"], query_accuracy["seed"] = model.upper(), seed
        enriched_item_frames.append(enriched)
        query_accuracy_frames.append(query_accuracy)
        geometry_group_frames.append(summarize_geometry_and_outcomes_by_group(enriched, topk=20))
        geometry_corr_frames.append(compute_geometry_output_correlations(enriched, topk=20))
        geometry_exposure_reg_frames.append(fit_geometry_exposure_models(enriched))
        geometry_accuracy_reg_frames.append(fit_geometry_accuracy_models(query_accuracy, enriched, topk=20, random_state=seed))

enriched_item_df = pd.concat(enriched_item_frames, ignore_index=True)
query_accuracy_df = pd.concat(query_accuracy_frames, ignore_index=True)
geometry_group_df = pd.concat(geometry_group_frames, ignore_index=True)
geometry_corr_df = pd.concat(geometry_corr_frames, ignore_index=True)
geometry_exposure_reg_df = pd.concat(geometry_exposure_reg_frames, ignore_index=True)
geometry_accuracy_reg_df = pd.concat(geometry_accuracy_reg_frames, ignore_index=True)
for frame in (geometry_group_df, geometry_corr_df, geometry_exposure_reg_df, geometry_accuracy_reg_df):
    frame["model"] = frame["model"].str.upper()

group_metrics = ["embedding_norm_mean", "centroid_distance_mean", "centroid_cosine_mean",
                 "exposure_share_pct", "within_group_coverage_pct", "recall@20"]
geometry_group_seed_summary = geometry_group_df.groupby(["model", "popularity_group"], as_index=False)[group_metrics].agg(["mean", "std"]).reset_index()
geometry_group_seed_summary.columns = ["_".join(str(x) for x in col if x).rstrip("_") if isinstance(col, tuple) else col for col in geometry_group_seed_summary.columns]
geometry_exposure_reg_seed_summary = geometry_exposure_reg_df.groupby(["model", "geometry_variable"], as_index=False).agg(
    geometry_beta_std_mean=("geometry_beta_std", "mean"), geometry_beta_std_sd=("geometry_beta_std", "std"),
    delta_r_squared_mean=("delta_r_squared_vs_frequency", "mean"), r_squared_mean=("r_squared", "mean"))
geometry_accuracy_reg_seed_summary = geometry_accuracy_reg_df.groupby(["model", "geometry_variable"], as_index=False).agg(
    geometry_beta_std_mean=("geometry_beta_std", "mean"), geometry_beta_std_sd=("geometry_beta_std", "std"),
    delta_log_loss_mean=("delta_log_loss_vs_frequency", "mean"), mcfadden_r_squared_mean=("mcfadden_r_squared", "mean"))

gap_source = geometry_group_df.pivot(index=["model", "seed"], columns="popularity_group",
    values=["embedding_norm_mean", "centroid_distance_mean", "exposure_share_pct", "recall@20"])
gap_source.columns = [f"{metric}_{group}" for metric, group in gap_source.columns]
geometry_bias_gap_df = gap_source.reset_index()
geometry_bias_gap_df["norm_gap_rare_minus_popular"] = geometry_bias_gap_df["embedding_norm_mean_rare"] - geometry_bias_gap_df["embedding_norm_mean_popular"]
geometry_bias_gap_df["centroid_gap_rare_minus_popular"] = geometry_bias_gap_df["centroid_distance_mean_rare"] - geometry_bias_gap_df["centroid_distance_mean_popular"]
geometry_bias_gap_df["exposure_gap_popular_minus_rare_pct"] = geometry_bias_gap_df["exposure_share_pct_popular"] - geometry_bias_gap_df["exposure_share_pct_rare"]
geometry_bias_gap_df["recall_gap_popular_minus_rare"] = geometry_bias_gap_df["recall@20_popular"] - geometry_bias_gap_df["recall@20_rare"]
geometry_bias_gap_df = geometry_bias_gap_df.merge(bias_df[["model", "seed", "gini_warm@20", "coverage_warm@20", "arp@20"]], on=["model", "seed"], validate="one_to_one")
gap_metrics = ["norm_gap_rare_minus_popular", "centroid_gap_rare_minus_popular", "exposure_gap_popular_minus_rare_pct",
               "recall_gap_popular_minus_rare", "gini_warm@20", "coverage_warm@20", "arp@20"]
geometry_bias_gap_seed_summary = geometry_bias_gap_df.groupby("model", as_index=False)[gap_metrics].agg(["mean", "std"]).reset_index()
geometry_bias_gap_seed_summary.columns = ["_".join(str(x) for x in col if x).rstrip("_") if isinstance(col, tuple) else col for col in geometry_bias_gap_seed_summary.columns]

geometry_group_df.to_csv(TABLES_DIR / "geometry_bias_group_summary.csv", index=False)
geometry_group_seed_summary.to_csv(TABLES_DIR / "geometry_bias_group_seed_summary.csv", index=False)
geometry_bias_gap_df.to_csv(TABLES_DIR / "geometry_bias_gap_summary.csv", index=False)
geometry_bias_gap_seed_summary.to_csv(TABLES_DIR / "geometry_bias_gap_seed_summary.csv", index=False)
geometry_corr_df.to_csv(STAT_TESTS_DIR / "geometry_output_correlations.csv", index=False)
geometry_exposure_reg_df.to_csv(STAT_TESTS_DIR / "geometry_exposure_regression.csv", index=False)
geometry_exposure_reg_seed_summary.to_csv(STAT_TESTS_DIR / "geometry_exposure_regression_seed_summary.csv", index=False)
geometry_accuracy_reg_df.to_csv(STAT_TESTS_DIR / "geometry_accuracy_logistic.csv", index=False)
geometry_accuracy_reg_seed_summary.to_csv(STAT_TESTS_DIR / "geometry_accuracy_logistic_seed_summary.csv", index=False)

print("=== A. GEOMETRY, EXPOSURE VÀ ACCURACY THEO NHÓM (MEAN 2 SEED) ===")
display(geometry_group_seed_summary[["model", "popularity_group", "embedding_norm_mean_mean",
    "centroid_distance_mean_mean", "exposure_share_pct_mean", "recall@20_mean"]])
print("=== B. GEOMETRY → EXPOSURE, ĐÃ KIỂM SOÁT FREQUENCY ===")
display(geometry_exposure_reg_seed_summary.query("geometry_variable != 'none'"))
print("=== C. GEOMETRY → HIT@20, ĐÃ KIỂM SOÁT FREQUENCY ===")
display(geometry_accuracy_reg_seed_summary.query("geometry_variable != 'none'"))
print("=== D. GEOMETRY GAP ĐẶT CẠNH BIAS GAP ===")
display(geometry_bias_gap_seed_summary)
```

## 6. Kết luận tự động, kiểm tra đầu ra và giới hạn diễn giải

Quy tắc đọc kết quả:

1. `freq ↔ rec_count` trả lời popularity trong train có đi kèm exposure đầu ra hay không.
2. `norm ↔ rec_count` tổng thể có thể bị nhiễu vì frequency đồng thời liên quan đến cả norm và recommendation count.
3. Với exposure, standardized beta và $\Delta R^2$ cho biết geometry bổ sung bao nhiêu thông tin sau khi giữ frequency cố định.
4. Với accuracy, logistic beta và mức giảm log-loss cho biết geometry có đi kèm xác suất hit@20 hay không.
5. Không suy diễn nhân quả: kết quả chỉ cho thấy các đặc điểm đi kèm nhau trong dữ liệu quan sát.
6. Cold item chỉ được dùng để audit hình học ở notebook 07, không trộn vào test warm hay hồi quy outcome chính.
7. Với chỉ hai seed, mean ± SD chỉ là kiểm tra ổn định ban đầu, không phải ước lượng bất định hoàn chỉnh.
